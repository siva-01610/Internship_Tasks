```python
#35
# [[1,99],[3,4]]
import copy

a = [[1, 2], [3, 4]]
b = copy.copy(a)
b[0][1] = 99
print(a)

#comment: A shallow copy creates a new outer list but keeps references to the same inner lists.
# Therefore changing an inner list through b also changes a.
```

```python
#36
# [[1, 2],[3, 4]]
import copy

a = [[1, 2], [3, 4]]
b = copy.deepcopy(a)
b[0][1] = 99
print(a)

#comment: deepcopy creates independent copies of the nested objects, so changing b does not affect a.
```

```python
#37
# [[10],[20, 30]]
import copy

a = [[10], [20]]
b = a.copy()
a[1].append(30)
print(b)

#comment: list.copy() performs a shallow copy, so b shares the inner lists with a.
```

```python
#38
# [[10],[20]]
import copy

a = [[10], [20]]
b = copy.deepcopy(a)
a[1].append(30)
print(b)

#comment: deepcopy creates separate inner lists, so changes to a do not affect b.
```

```python
#39
# False,True
import copy

a = [[1, 2], [3, 4]]
b = copy.copy(a)
print(a is b)
print(a[0] is b[0])

#comment: Shallow copying creates a new outer list but keeps the same inner list objects.
```

```python
#44
# (1, 2)
def test(*args):
    print(args)

test(1, 2)

#comment: *args collects all positional arguments into a tuple.
```

```python
#45
# tuple
def test(*args):
    print(type(args))

test("a", "b", "c")

#comment: Positional arguments collected by *args are stored as a tuple.
```

```python
#46
# {'name':'Ram','age':25}
def test(**kwargs):
    print(kwargs)

test(name="Ram", age=25)

#comment: **kwargs collects keyword arguments into a dictionary.
```

```python
#47
# <class 'dict'>
def test(**kwargs):
    print(type(kwargs))

test(a=10, b=20)

#comment: Keyword arguments collected using **kwargs are stored in a dictionary.
```
