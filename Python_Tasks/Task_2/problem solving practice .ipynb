```python
# Q1. Find the Shortest Word in a Sentence
# Write a Python program to find the shortest word in a given sentence.
# Example input:
# Python makes coding easy
# Expected output:
# easy

sentence = input("Enter a sentence")
words = sentence.split()
short = min(words, key=len)
print(short)
```

```python
# Q2. Find the Largest Number in a List
# Write a Python program to find the largest number in a list without using the built-in max() function.
# Example input:
# [12, 45, 7, 89, 23]
# Expected output:
# 89
l=list(map(int, input("Enter numbers").split()))
max=l[0]
for i in l:
    if i>max:
        max=i
print(max)
```

```python
# Q3. Find the Smallest Number in a List
# Write a Python program to find the smallest number in a list without using the built-in min() function.
# Example input:
# [12, 45, 7, 89, 23]
# Expected output:
# 7
l=list(map(int, input("Enter numbers").split()))
min=l[0]
for i in l:
    if i<min:
        min=i
print(min)
```

```python
# Q4. Find the First Longest Word When Ties Occur
# Write a Python program to find the first longest word in a sentence. If multiple words have the same maximum length, return the one that appears first.
# Example input:
# Python makes coding useful
# Expected output:
# Python
sentence=input("enter sentence").split()
largest=sentence[0]
for word in sentence:
    if len(word)>len(largest):
        largest=word
print(largest)
```

```python
# Q6. Count Vowels in a String
# Write a Python program to count the total number of vowels in a string. Consider both uppercase and lowercase vowels.
# Example input:
# Education
# Expected output:
# 5
s=input("enter word")
vowel_count=0
for i in s:
    if i in 'aeiouAAEIOU':
        vowel_count+=1
print(vowel_count)
```

```python
# Q7. Count Consonants in a String
# Write a Python program to count the consonants in a string. Ignore spaces, digits, and special characters.
# Example input:
# Python 123!
# Expected output:
# 5
s=input("enter word")
conso_count=0
for i in s:
    if i.isalpha() and i not in 'aeiouAAEIOU':
        conso_count+=1
print(conso_count)
```

```python
# Q8. Find the Most Frequent Character
# Write a Python program to find the character that appears most frequently in a string. If multiple characters have the same highest frequency, return the first one encountered. Ignore spaces.
# Example input:
# banana
# Expected output:
# a
s = input("Enter a string: ")
s = s.replace(" ", "")
max_count = 0
for i in s:
    count = s.count(i)
    if count > max_count:
        max_count = count
        ch = i

print(ch)
```

```python
# Q9. Find the Least Frequent Character
# Write a Python program to find the character that appears least frequently in a string. If multiple characters have the same lowest frequency, return the first one encountered. Ignore spaces.
# Example input:
# success
# Expected output:
# u
s = input("Enter a string: ")
s = s.replace(" ", "")
min_count = len(s)
for i in s:
    count = s.count(i)
    if count < min_count:
        min_count = count
        ch = i
print(ch)
```

```python
# Q10. Find All Characters Appearing More Than Once
# Write a Python program to identify all characters that appear more than once in a string. Display each repeated character only once, in the order of its first appearance. Ignore spaces.
# Example input:
# programming
# Expected output:
# ['r', 'g', 'm']
word=input("enter word")
l=[]
for i in word:
    if word.count(i)>1 and i not in l:
        l.append(i)
print(l)
```

```python
# Q11. Create a Dictionary of Words and Their First Characters
# Write a Python program to create a dictionary in which each word is a key and its first character is the value.
# Example input:
# Python Data Engineering
# Expected output:
# {'Python': 'P', 'Data': 'D', 'Engineering': 'E'}
s = input("Enter a sentence: ")
words = s.split()
d = {}
for i in words:
    d[i] = i[0]
print(d)
```

```python
# Q13. Create a Dictionary of Words and Their Uppercase Versions
# Write a Python program to create a dictionary in which each word is a key and its uppercase version is the value.
# Example input:
# data science python
# Expected output:
# {'data': 'DATA', 'science': 'SCIENCE', 'python': 'PYTHON'}
s = input("Enter a sentence: ")
words = s.split()
d = {}
for i in words:
    d[i] = i.upper()
print(d)
```

```python
# Q14. Create a Dictionary of Numbers and Their Squares
# Write a Python program to create a dictionary containing each number in a list as a key and its square as the value.
# Example input:
# [2, 3, 4, 5]
# Expected output:
# {2: 4, 3: 9, 4: 16, 5: 25}
l = [2, 3, 4, 5]
d = {}

for i in l:
    d[i] = i ** 2

print(d)
```

```python
# Q15. Create a Dictionary of Numbers and Their Even/Odd Status
# Write a Python program to create a dictionary containing each number as a key and "Even" or "Odd" as its value.
# Example input:
# [1, 2, 3, 4, 5]
# Expected output:
# {1: 'Odd', 2: 'Even', 3: 'Odd', 4: 'Even', 5: 'Odd'}
l = [1, 2, 3, 4, 5]
d = {}

for i in l:
    if i % 2 == 0:
        d[i] = "Even"
    else:
        d[i] = "Odd"

print(d)
```

```python
# Q19. Count Words Whose Length Is Greater Than 4
# Write a Python program to count the words containing more than four characters in a sentence.
# Example input:
# Python makes problem solving interesting
# Expected output:
# 4
s = input("Enter a sentence: ")
words = s.split()
count = 0

for i in words:
    if len(i) > 4:
        count += 1

print(count)
```
